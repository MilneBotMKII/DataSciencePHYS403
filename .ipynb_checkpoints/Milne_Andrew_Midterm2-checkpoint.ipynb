Part 1.)

```python
#our model function needs a few things:
#We need to solve for the flat background. Some estimator will be used for this(minimize a flat line to the data)
#We need to have a small gaussian for the signal peak. Probbaly at 80 GeV but we'll see what it optimizes to
#So this will be a 4 parameter model. Background, Gaussian center, Gaussian width, Gaussian amplitude
```

```python
#Copying some code from HW4

def Model(x, Background, Location, Width, Amplitude):
    #We have a constant flat background, and a gaussian added on top with a width, amplitude, and location
    return(Background + Amplitude * np.exp(-(x - Location)**2 / (2 * Width**2)))

#We define our priors. I'll do a Jefferys prior for the width and uniform for the others
#I'll do a log(Prior) function since I am working in LogPosterior space
def logPrior(Background, Location, Width, Amplitude):
    a=1 #parameter for modified jefferys prior
    logWidthPrior = -np.log(Width+a) - np.log(np.log((a+np.max(WidthList))/a)) #Jefferys for gaussian width. Modified by a
    logBackgroundPrior = 0 #uniform in prior means 0 in logprior. Or rather a constant but we can set to 0 safely
    logLocationPrior = 0
    logAmplitudePrior = 0
    return(logWidthPrior + logBackgroundPrior + logLocationPrior + logAmplitudePrior) #log space so multiplying priors is adding logpriors

def LogPosterior(Data, Background, Location, Width, Amplitude):    
    
    sigma = np.sqrt(Data['counts'] + 1) #+1 just avoids any 0 cases in case I mess up lol. Should be small compared to everything else
    logL = np.sum(-0.5 * np.log(2*np.pi*sigma**2)
                   - (Data['counts'] - Model(Data['mass'], Background, Location, Width, Amplitude))**2 / (2*sigma**2))
    return(logL + logPrior(Background, Location, Width, Amplitude))

```

```python
#blarg need to keep this pretty course for it to run in reasonable time. At least for testing it out
BackgroundList = np.linspace(80, 130, 20)
LocationList = np.linspace(np.min(Data['mass']), np.max(Data['mass']), 50)
WidthList = np.logspace(-0.5, 1, 20) #concentrate more on small sigma
AmplitudeList = np.linspace(-50, 100, 30)

B, L, W, A = np.meshgrid(BackgroundList, LocationList, WidthList, AmplitudeList) #makes the 2D grid
theta = np.column_stack((B.ravel(), L.ravel(), W.ravel(), A.ravel())) #and flattens it out
```

```python
#Now we have our parameter priors as well as our test values for those parameters. Plug them in
logPDFList=[]
for i in range(len(theta)):
    logPDFList.append(LogPosterior(Data, theta[i][0], theta[i][1], theta[i][2], theta[i][3]))
logPDFList = np.array(logPDFList)
logPDFShift = logPDFList - np.max(logPDFList) #shifts by a constant in log space so multiplies by a constant in P space
#Just to move us away from 0
```

```python
#Now we have all our logPosteriors, we can find the optimal parameters
#This probably needs marginalization over the data
PDF = np.exp(logPDFShift)
def HDP(x, PDF, alpha):
    dx = np.diff(x)
    dx = np.append(dx, dx[-1]) #just to make same length. And leaving general for future homeworks
    #OK so the way I'm going about this is to sort by probability
    idx = np.argsort(PDF)[::-1] #in this case first is highest likelyhood, last is least likely
    temp = []
    cumulative = 0
    for i in idx: #ok so basically we sort the pdf by decreasing probability, and scan in new values if we haven't hit alpha
        cumulative += PDF[i] * dx[i]
        temp.append(i)
        if cumulative >= alpha:
            break
    temp = np.array(temp)
    #HPD is min/max value
    HDPmin = x[temp].min()
    HDPmax = x[temp].max()
    return(HDPmin, HDPmax) #This is probably overkill but this does account for everything

alpha=0.68

#Background estimator marginalization
P_Background = []
for value in BackgroundList:
    Mask = (B.ravel() == value) #masks all inputs with a specific background value
    #the ravel is what we used to construct theta
    P_Background.append(np.sum(PDF[Mask])) #sums over all those probabilities
P_Background = np.array(P_Background)
dx = BackgroundList[1] - BackgroundList[0]
P_Background /= np.sum(P_Background * dx) #normalize

MPV_Background = BackgroundList[np.argmax(P_Background)]
HDPmin, HDPmax = HDP(BackgroundList, P_Background, alpha=alpha)
plt.plot(BackgroundList, P_Background, label = 'Marginalized PDF')
plt.vlines(MPV_Background, np.min(P_Background), np.max(P_Background), 
           label = 'MPV: '+str(round(MPV_Background, ndigits=3)), color='red', linestyle='--')

mask = (BackgroundList >= HDPmin) & (BackgroundList <= HDPmax)
plt.fill_between(BackgroundList[mask], P_Background[mask], alpha=0.5, color='red', 
                 label = str(100*alpha) + '% credible: ' + str(round(HDPmin, ndigits=3)) + ' to ' + str(round(HDPmax, ndigits=3)))

plt.xlabel('Background')
plt.ylabel('P(Background|Data)')
plt.title('Marginalized Posterior for Background')
plt.legend()
plt.show()

#Location estimator marginalization
P_Location = []
for value in LocationList:
    Mask = (L.ravel() == value) #masks all inputs with a specific background value
    #the ravel is what we used to construct theta
    P_Location.append(np.sum(PDF[Mask])) #sums over all those probabilities
P_Location = np.array(P_Location)
dx = LocationList[1] - LocationList[0]
P_Location /= np.sum(P_Location * dx) #normalize

MPV_Location = LocationList[np.argmax(P_Location)]
HDPmin, HDPmax = HDP(LocationList, P_Location, alpha=alpha)
plt.plot(LocationList, P_Location, label = 'Marginalized PDF')
plt.vlines(MPV_Location, np.min(P_Location), np.max(P_Location), 
           label = 'MPV: '+str(round(MPV_Location, ndigits=3)), color='red', linestyle='--')

mask = (LocationList >= HDPmin) & (LocationList <= HDPmax)
plt.fill_between(LocationList[mask], P_Location[mask], alpha=0.5, color='red', 
                 label = str(100*alpha) + '% credible: ' + str(round(HDPmin, ndigits=3)) + ' to ' + str(round(HDPmax, ndigits=3)))

plt.xlabel('Location')
plt.ylabel('P(Location|Data)')
plt.title('Marginalized Posterior for Location')
plt.legend()
plt.show()

#Width estimator marginalization
P_Width = []
for value in WidthList:
    Mask = (W.ravel() == value) #masks all inputs with a specific background value
    #the ravel is what we used to construct theta
    P_Width.append(np.sum(PDF[Mask])) #sums over all those probabilities
P_Width = np.array(P_Width)
dx = np.diff(WidthList)
dx=np.append(dx, dx[-1]) #uhh this isn't great but idk how better to handle
P_Width /= np.sum(P_Width * dx) #normalize

MPV_Width = WidthList[np.argmax(P_Width)]
HDPmin, HDPmax = HDP(WidthList, P_Width, alpha=alpha)
plt.plot(WidthList, P_Width, label = 'Marginalized PDF')
plt.vlines(MPV_Width, np.min(P_Width), np.max(P_Width), 
           label = 'MPV: '+str(round(MPV_Width, ndigits=3)), color='red', linestyle='--')

mask = (WidthList >= HDPmin) & (WidthList <= HDPmax)
plt.fill_between(WidthList[mask], P_Width[mask], alpha=0.5, color='red', 
                 label = str(100*alpha) + '% credible: ' + str(round(HDPmin, ndigits=3)) + ' to ' + str(round(HDPmax, ndigits=3)))

plt.xlabel('Width')
plt.ylabel('P(Width|Data)')
plt.title('Marginalized Posterior for Width')
plt.legend()
plt.show()

#Amplitude estimator marginalization
P_Amplitude = []
for value in AmplitudeList:
    Mask = (A.ravel() == value) #masks all inputs with a specific background value
    #the ravel is what we used to construct theta
    P_Amplitude.append(np.sum(PDF[Mask])) #sums over all those probabilities
P_Amplitude = np.array(P_Amplitude)
dx = AmplitudeList[1] - AmplitudeList[0]
P_Amplitude /= np.sum(P_Amplitude * dx) #normalize

MPV_Amplitude = AmplitudeList[np.argmax(P_Amplitude)]
HDPmin, HDPmax = HDP(AmplitudeList, P_Amplitude, alpha=alpha)
plt.plot(AmplitudeList, P_Amplitude, label = 'Marginalized PDF')
plt.vlines(MPV_Amplitude, np.min(P_Amplitude), np.max(P_Amplitude), 
           label = 'MPV: '+str(round(MPV_Amplitude, ndigits=3)), color='red', linestyle='--')

mask = (AmplitudeList >= HDPmin) & (AmplitudeList <= HDPmax)
plt.fill_between(AmplitudeList[mask], P_Amplitude[mask], alpha=0.5, color='red', 
                 label = str(100*alpha) + '% credible: ' + str(round(HDPmin, ndigits=3)) + ' to ' + str(round(HDPmax, ndigits=3)))

plt.xlabel('Amplitude')
plt.ylabel('P(Amplitude|Data)')
plt.title('Marginalized Posterior for Amplitude')
plt.legend()
plt.show()
```

```python
#A frequentist approach. For this we will minimize chi2
def Chi2(Data, Background, Location, Width, Amplitude):
    Chi2=np.sum((Data['counts'] - Model(Data['mass'], Background, Location, Width, Amplitude))**2 / np.sqrt(Data['counts'])**2)
    return(Chi2)

def Chi2Minimizer(theta, Data):
    B, L, W, A = theta
    return Chi2(Data, B, L, W, A)

chi2minimized=scipy.optimize.minimize(Chi2Minimizer, x0=[100, 80, 2, 30], args=(Data,)) #some initial guesses
BackgroundOptimized, LocationOptimized, WidthOptimized, AmplitudeOptimized = chi2minimized.x
chi2optimizedminimumBest = chi2minimized.fun
print('Optimized Background:', BackgroundOptimized)
print('Optimized Location:', LocationOptimized)
print('Optimized Width:', WidthOptimized)
print('Optimized Amplitude:', AmplitudeOptimized)
print('Chi2 of Minimized Result:', chi2optimizedminimumBest)

#Let's plot the most optimal value
plt.errorbar(Data['mass'], Data['counts'], yerr=np.sqrt(Data['counts']), label='Data', linestyle='', fmt='.')
plt.plot(Data['mass'], Model(Data['mass'], BackgroundOptimized, LocationOptimized, WidthOptimized, AmplitudeOptimized), label='Best model parameters')
plt.legend()
plt.xlabel('Mass(GeV)')
plt.ylabel('Counts')
plt.title('Optimized Parameters')
plt.show()

#This does look good, but I will say I had to kinda force the initial guess values to be close to the results from 
#Bayesian results. So, I want to also show a less prompted result

chi2minimized=scipy.optimize.minimize(Chi2Minimizer, x0=[80, 70, 0.1, 10], args=(Data,)) #some initial guesses
BackgroundOptimized2, LocationOptimized2, WidthOptimized2, AmplitudeOptimized2 = chi2minimized.x
chi2optimizedminimumOff = chi2minimized.fun
print('Optimized Background:', BackgroundOptimized2)
print('Optimized Location:', LocationOptimized2)
print('Optimized Width:', WidthOptimized2)
print('Optimized Amplitude:', AmplitudeOptimized2) #well this doesn't match what's being plotted real well...
print('Chi2 of Minimized Result:', chi2optimizedminimumOff)
plt.errorbar(Data['mass'], Data['counts'], yerr=np.sqrt(Data['counts']), label='Data', linestyle='', fmt='.')
plt.plot(Data['mass'], Model(Data['mass'], BackgroundOptimized2, LocationOptimized2, WidthOptimized2, AmplitudeOptimized2), label='Best model parameters')
plt.legend()
plt.xlabel('Mass(GeV)')
plt.ylabel('Counts')
plt.title('Chi2 Unprompted')
plt.show()

#As can be seen, the optimizer struggles with moving the location parameter in particular
```

```python
#We can look at the chi2 of a just background model.
def Chi2Background(Data, Background):
    Chi2=np.sum((Data['counts'] - Background)**2 / np.sqrt(Data['counts'])**2)
    return(Chi2)

def Chi2Minimizer(B, Data):
    return Chi2Background(Data, B)

chi2minimized=scipy.optimize.minimize(Chi2Minimizer, x0=[10], args=(Data,)) #some initial guesses
BackgroundOnlyOptimized = chi2minimized.x
chi2optimizedminimumFlat = chi2minimized.fun
print('Optimized Background:', BackgroundOnlyOptimized[0])
print('Chi2 of Minimized Result:', chi2optimizedminimumFlat)

plt.errorbar(Data['mass'], Data['counts'], yerr=np.sqrt(Data['counts']), label='Data', linestyle='', fmt='.')
plt.plot(Data['mass'], np.zeros(np.shape(Data['mass'])) + BackgroundOnlyOptimized, label='Background Only')
plt.legend()
plt.xlabel('Mass(GeV)')
plt.ylabel('Counts')
plt.title('Just Background')
plt.show()
```

```python
#With the chi2 of the best fit, as well as just the background, we can say the significance of the singal
#We can say the deltaChi2 between the null result(just background) and the optimized signal(at ~80 GeV, see below)
deltachi2=chi2optimizedminimumFlat - chi2optimizedminimumBest
print('Deltachi2:', round(deltachi2, ndigits=5))
#We expect this to be on a chi2 distribution of 3 degrees of freedom(the 3 gaussian parameters)
#therefore we can place this deltachi2 on the 3DoF distribution and show it's significance
p_value = 1 - scipy.stats.chi2.cdf(deltachi2, df=3)
print('p value:', p_value)
#we can convert this to a 'sigma' detection through the survival function
print('Sigma detection:', scipy.stats.norm.isf(p_value))
#so not quite 5 sigma like we would want for a particle detection. But strong enough we might not have an oops-leon
```
